## Exploratory Data Analysis (EDA) of House Prices

The objective of this project is to understand the factors that influence house prices.
Exploratory Data Analysis is used to study data distribution, relationships between variables,
and patterns that affect the target variable SalePrice.


```python
# Data manipulation and analysis
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python

# Load the dataset
df = pd.read_csv("housings_data.csv")

df.head()
```

The dataset used in this project is the Ames Housing dataset.
It contains information related to house size, quality, rooms, year built, and sale price.
The target variable for this analysis is SalePrice.

<h1 style="text-align:center; font-size:32px;">##UNIVARIATE ANALYSIS##</h1>




Univariate analysis means analyzing one variable at a time to understand its distribution, frequency, and pattern without comparing it with other variables.

```python
plt.figure(figsize=(6,4))
sns.histplot(df['SalePrice'], kde=True , color='green')
plt.title("Univariate Analysis of SalePrice")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.show()

```

**Observation:**
The SalePrice distribution is right-skewed.
Most houses fall in the mid-price range, while a few houses have very high prices.


```python
plt.figure(figsize=(4,5))
sns.boxplot(y=df['SalePrice'], color='red')
plt.title("Outliers in SalePrice")
plt.show()
```

**Outlier Analysis:**
Some houses have extremely high prices compared to others.
These outliers were identified using boxplots and handled carefully to avoid data loss.


```python
df['SalePrice_log'] = np.log(df['SalePrice'])
```

```python
sns.histplot(df['SalePrice_log'], kde=True,color='blue')
plt.title("Log Transformed SalePrice")
plt.show()
```

**Observation:**
After applying log transformation, the price distribution becomes more balanced.
This reduces the effect of extreme values and makes the data easier to analyze.
Log transformation helps in handling skewness caused by high-priced houses.


<h1 style="text-align:center; font-size:32px;">##BIVARIATE ANALYSIS##</h1>

Bivariate analysis studies the relationship between two variables to understand how one variable affects another.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='MSSubClass', y='SalePrice', data=df,color='purple')
plt.title("MSSubClass vs SalePrice")
plt.xlabel("MSSubClass")
plt.ylabel("Sale Price")
plt.xticks(rotation=45)
plt.show()
```

**Observation:**
Different dwelling types have different price ranges.
Some house types are more expensive, while others are more affordable.
This shows that dwelling type also influences house prices.


```python
avg_price = df.groupby('MSSubClass')['SalePrice'].mean().sort_values()

avg_price.plot(kind='bar', figsize=(10,4),)
plt.title("Average SalePrice by MSSubClass")
plt.xlabel("MSSubClass")
plt.ylabel("Average Sale Price")
plt.show()
```

**Observation:**
This bar chart shows the average sale price for each MSSubClass.
Some dwelling types have a higher average price, while others are more affordable.
This confirms that house type has a clear impact on pricing.


```python
plt.figure(figsize=(4,4))
sns.scatterplot(x='GrLivArea', y='SalePrice', data=df,color='green')
plt.title("Living Area vs SalePrice")
plt.xlabel("Living Area")
plt.ylabel("SalePrice")
plt.show()
```

**Observation:**
Living area shows a strong positive relationship with SalePrice.
As the size of the house increases, the price also increases.


```python
plt.figure(figsize=(6,4))
sns.boxplot(x='OverallQual', y='SalePrice', data=df,color='red')
plt.title("Overall Quality vs SalePrice")
plt.show()
```

**Observation:**
Overall quality has a strong impact on house prices.
Higher quality houses are significantly more expensive.


```python
plt.figure(figsize=(6,4))
sns.boxplot(x='GarageCars', y='SalePrice', data=df,color='green')
plt.title("Garage Capacity vs SalePrice")
plt.show()
```

**Observation:**
Houses with more garage capacity tend to have higher prices.
Garage space adds convenience and increases the value of the property.


```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='YearBuilt', y='SalePrice', data=df,color='brown')
plt.title("Year Built vs SalePrice")
plt.show()
```

**Observation:**
Newer houses generally have higher sale prices compared to older houses.
This shows that modern construction and newer designs attract higher value.


```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='TotalBsmtSF', y='SalePrice', data=df)
plt.title("Basement Area vs SalePrice")
plt.xlabel("Total Basement Area")
plt.ylabel("Sale Price")
plt.show()
```

**Observation:**
Basement area shows a positive relationship with sale price.
Houses with larger basement areas generally have higher prices.
This indicates that additional usable space increases property value.


<h1 style="text-align:center; font-size:32px;">##MULTIVARIATE ANALYSIS##</h1>

Multivariate analysis means analyzing more than two variables together to understand their combined effect on the target variable (SalePrice).

```python
multi_df = df[
    [
        'SalePrice',
        'OverallQual',
        'GrLivArea',
        'TotalBsmtSF',
        'GarageCars',
        'YearBuilt',
        'FullBath',
        'BedroomAbvGr'
    ]
]


```

```python
sns.pairplot(multi_df)
plt.show()
```

**Observation:**
This pair plot shows relationships between multiple numerical variables at the same time.
OverallQual and GrLivArea show the strongest relationship with SalePrice.
Other features like TotalBsmtSF and GarageCars also support higher prices.
This plot gives an overall view of how multiple factors together affect house prices.




```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    x='GrLivArea',
    y='SalePrice',
    hue='OverallQual',
    palette='coolwarm',
    data=multi_df
)
plt.title("Living Area vs SalePrice by Overall Quality")
plt.xlabel("Living Area (sq ft)")
plt.ylabel("Sale Price")
plt.show()
```

**Observation:**
This graph shows the combined effect of living area and overall quality on house prices.
For the same house size, higher quality houses have higher prices.
This confirms that both size and quality together play a major role in pricing.


```python
plt.figure(figsize=(6,4))
sns.boxplot(
    x='BedroomAbvGr',
    y='SalePrice',
    hue='FullBath',
    palette='pastel',
    data= multi_df
)
plt.title("Bedrooms vs SalePrice by Bathrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Sale Price")
plt.show()
```

**Observation:**
This plot shows the combined effect of bedrooms and bathrooms on house prices.
For the same number of bedrooms, houses with more bathrooms generally have higher prices.
This indicates that comfort-related features together influence pricing.


```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    x='YearBuilt',
    y='SalePrice',
    hue='GarageCars',
    palette='viridis',
    data=multi_df
)
plt.title("Year Built vs SalePrice by Garage Capacity")
plt.xlabel("Year Built")
plt.ylabel("Sale Price")
plt.show()
```

**Observation:**
This graph shows that newer houses generally have higher prices.
Among houses built in the same year, those with larger garage capacity tend to be more expensive.
This highlights the combined impact of house age and garage space on pricing.


```python
num_df = df.select_dtypes(include='number')
corr_matrix = num_df.corr()

plt.figure(figsize=(5,3))
sns.heatmap(corr_matrix, cmap='coolwarm', center=0)
plt.title("Correlation Heatmap of All Numerical Features")
plt.show()
```

**Observation:**
This heatmap represents correlations among all numerical features in the dataset.
It provides an unbiased view of which variables are strongly or weakly related to SalePrice.
Strong correlations are observed with quality and size-related features.




```python
plt.figure(figsize=(8,6))
sns.heatmap(
    multi_df.corr(),
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)
plt.title("Multivariate Correlation Heatmap")
plt.show()
```

**Observation:**
This heatmap shows the correlation between selected important variables.
OverallQual and GrLivArea have the strongest positive correlation with SalePrice.
Other features such as TotalBsmtSF and GarageCars also show good correlation.
This heatmap confirms the results observed in previous graphs.
